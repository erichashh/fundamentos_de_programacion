Ejercicio 10: suma acumulada con while

Pedimos números y los vamos sumando. Cuando el usuario ingresa 0, el ciclo termina y ese 0 no se suma.

```python
suma = 0
numero = int(input("dame un numero (0 para terminar): "))

while numero != 0:
    suma = suma + numero
    numero = int(input("Da un numero (0 para terminar): "))

print("La suma de los numeros es:", suma)
```

Ejercicio 12: suma de pares del 1 al 20

Recorremos del 1 al 20 y con continue nos saltamos los impares, sumando solo los pares.

```python
suma = 0

for numero in range(1, 21):
    if numero % 2 != 0:
        continue
    suma = suma + numero

print("La suma de los números pares del 1 al 20 es ", suma)
```

Ejercicio 13: suma de una lista con función

Creamos una función que recorre la lista y va sumando cada elemento, sin usar sum().

```python
def sumar_lista(lista):
    total = 0
    for elemento in lista:
        total = total + elemento
    return total

numeros = [10, 20, 30, 40]
resultado = sumar_lista(numeros)

print("La suma de los elementos es:", resultado)
```

Ejercicio 14: máximo de una lista con función

La función va comparando los elementos de la lista para encontrar el más grande, sin usar max().

```python
def encontrar_maximo(lista):
    maximo = lista[0]
    for elemento in lista:
        if elemento > maximo:
            maximo = elemento
    return maximo

numeros = [7, 3, 9, 2]
resultado = encontrar_maximo(numeros)

print("El elemento maximo es:", resultado)
```

Ejercicio 15: suma de la diagonal principal de una matriz

La diagonal principal son los elementos donde la fila y la columna tienen el mismo índice.

```python
matriz = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]

suma = 0
for i in range(len(matriz)):
    suma = suma + matriz[i][i]

print("La suma de la diagonal es:", suma)
```

Ejercicio 16: promedio con función que no regresa valor

La función calcula el promedio y lo imprime directamente adentro, sin usar return.

```python
def mostrar_promedio(lista):
    suma = 0
    for elemento in lista:
        suma = suma + elemento
    promedio = suma / len(lista)
    print(f"promedio es: {promedio:.2f}")

numeros = [8, 9, 10, 7]
mostrar_promedio(numeros)
```

Ejercicio 18: factorial con función recursiva

La función se llama a sí misma hasta llegar al caso base (cuando n es 0 o 1).

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

numero = 5
resultado = factorial(numero)

print(f"El factorial de {numero} es:", resultado)
```

Ejercicio 21: diccionario de calificaciones y promedios

El diccionario guarda una lista de calificaciones por alumno. Recorremos el diccionario y calculamos el promedio de cada uno.

```python
calificaciones = {"Ana": [8, 9, 10], "Beto": [7, 7, 8]}

for alumno in calificaciones:
    lista = calificaciones[alumno]
    suma = 0
    for nota in lista:
        suma = suma + nota
    promedio = suma / len(lista)
    print(f"{alumno}: {promedio:.2f}")
```
