Ejercicio 3 detectar el error
El error es que  12 se resta directamente como dinero, en lugar de representar el 12% del precio.

Lo corrijo:

```python
import pdb

precio=45
descuento=12

pdb.set_trace()
descuento = precio * (descuento / 100)
total = precio - descuento

print(f"Total: {total:.2f}")
```

El total nos queda que es: 39.60
